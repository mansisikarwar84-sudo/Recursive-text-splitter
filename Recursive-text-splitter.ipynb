```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
unstructured_data = "Generative AI is a subset of deep learning. It uses transformer models to create text. Students love learning LangChain because it makes things easy."
```

```python
# Splitter Configuration 
# Notice: We don't provide an explicit separator here; it uses the default hierarchical list: ["\n\n", "\n", " ", ""]
recursive_splitter = RecursiveCharacterTextSplitter(
    chunk_size=40,   # Keeping the same 40 character limit
    chunk_overlap=5  # A slight overlap to maintain context between chunks
)
```

```python
print("=== CASE 3: Recursive Character Splitter (Problem Solved) ===")
chunks_recursive = recursive_splitter.split_text(unstructured_data)

for i, chunk in enumerate(chunks_recursive):
    print(f"Chunk {i+1} (Size {len(chunk)}): {chunk}")

# HOW THE OUTPUT WILL LOOK:
# It will smartly fall back to splitting based on full-stops and spaces.
# Chunk 1: Generative AI is a subset of deep
# Chunk 2: deep learning. It uses transformer
```
